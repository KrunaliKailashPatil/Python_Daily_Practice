```python
# Q2. Add a New Product
products = {
    "Mouse":600,
    "Laptop":7000,
    "Keyboard":3000
}
products["Monitor"] = 2500
print(products)
```

```python
# Q3. Update Employee Salary
salary = {
    "Rahul":80000,
    "Akshat":60000,
    "Prem":5000
}
salary["Prem"] = 50000
print(salary)
```

```python
# Q4. Find Total Sales
sales = {
    "Shoes":1000,
    "Bags":3000,
    "Watches":4000
}
total_sales = sum(sales.values())
print(sales)
print(total_sales)
```

```python
# Q5. Find the Highest Scorer
student_marks = {
    "Anjali":85,
    "Rahul": 92,
    "Priya": 78,
    "Amit": 88,
    "Sneha": 95
}

top_student = max(student_marks,key=student_marks.get)
print(student_marks)
print(top_student)
print(student_marks[top_student])
```

```python
# Q6. Count Product Categories
categories = [
    "Clothing",
    "Shoes",
    "Clothing",
    "Electronics",
    "Shoes",
    "Clothing"
]
category_count = {}

for category in categories:
    category_count[category] = category_count.get(category,0) + 1

print(category_count)
```

```python
# Q7. Find Customers Who Have Not Paid
payment = {
    "Amit": "Paid",
    "Priya": "Pending",
    "Rahul": "Pending",
    "Sneha": "Paid",
    "Karan": "Pending"
}
for customer,status in payment.items():
    if status == "Pending":
        print(customer)
```

```python
# Q8. Calculate Average Marks
student_marks = {
    "Anjali":85,
    "Rahul": 92,
    "Priya": 78,
    "Amit": 88,
    "Sneha": 95
}
avg_marks = sum(student_marks.values()) / len(student_marks)
print(avg_marks)
```

```python
# Q9. Filter High-Value Customers
customer_spending = {
    "Amit": 2500,
    "Priya": 7000,
    "Rahul": 5500,
    "Sneha": 3200,
    "Karan": 9000
}
for customer,amount in customer_spending.items():
    if amount > 5000:
        print(customer,":",amount)
```

```python
# Q10. Create a Discounted Price Dictionary
product_prices = {
    "Shoes": 2000,
    "Bag": 1500,
    "Watch": 3000
}

discounted_prices = {}
for product, price in product_prices.items():
    discounted_prices[product] = price * 0.90

print(discounted_prices)
```
