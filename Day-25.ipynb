```python
# Question 1: Find Customers with More Than One Purchase
purchases = {
    "C101": 3,
    "C102": 1,
    "C103": 5,
    "C104": 2,
    "C105": 1
}
for customer,order in purchases.items():
    if order > 2:
        print(customer,":",order)
```

```python
# Question 2: Replace Negative Values with Zero
profit = {
    "Monday": 500,
    "Tuesday": -200,
    "Wednesday": 300,
    "Thursday": -100
}
for day in profit:
    if profit[day] < 0:
        profit[day]=0
print(profit)
```

```python
# Question 3: Calculate Average Product Rating
ratings = {
    "Laptop": 4.5,
    "Mouse": 4.0,
    "Keyboard": 3.5,
    "Monitor": 5.0
}
avgrage = sum(ratings.values())/len(ratings)
print(avgrage)
```

```python
# Question 4: Find Customers Who Have Not Placed an Order
orders = {
    "C101": 5,
    "C102": 0,
    "C103": 3,
    "C104": 0,
    "C105": 2
}
for id,order in orders.items():
    if order == 0:
        print(id,":",order)
```

```python
# Question 5: Calculate Discounted Prices
prices = {
    "Shirt": 1000,
    "Jeans": 2000,
    "Shoes": 3000
}
discount = {}
for product,price in prices.items():
    discount[product]=price*0.90
print(discount)
```

```python
# Question 6: Find the Department with the Most Employees
employees = {
    "Amit": "IT",
    "Priya": "HR",
    "Rahul": "IT",
    "Sneha": "Finance",
    "Neha": "IT",
    "Karan": "HR"
}
department_count = {}
for department in employees.values():
    department_count[department]=department_count.get(department,0)+1
print(department_count)
```

```python
# Question 8: Merge Two Dictionaries and Add Matching Values
branch_a = {
    "Laptop": 50000,
    "Mouse": 2000,
    "Keyboard": 3000
}

branch_b = {
    "Laptop": 40000,
    "Mouse": 1500,
    "Monitor": 10000
}
total_sales = branch_a.copy()
for product, amount in branch_b.items():
    total_sales[product] = total_sales.get(product, 0) + amount
print(total_sales)
```

```python
# Question 9: Find the Product with the Highest Revenue
revenue = {
    "Laptop": 150000,
    "Mobile": 200000,
    "Tablet": 80000,
    "Smartwatch": 60000
}
top_product=max(revenue,key=revenue.get)
print(revenue)
print("Top product:",top_product)
print("revenue:",revenue[top_product])
```

```python
# Question 10: Convert Marks into Grades
result = {
    "Anjali": 95,
    "Rahul": 82,
    "Meera": 68,
    "Karan": 35
}
grade={}
for student,marks in result.items():
    if marks >= 90:
        grade[student,marks]='A'
    elif marks >= 75:
        grade[student,marks]='B'
    elif marks >=50:
        grade[student,marks]='C'
    else:
        grade[student,marks]='Fail'
print(grade)
```

```python
# Question 11: Find the Most Frequently Sold Product
quantity_sold = {
    "Laptop": 12,
    "Mouse": 45,
    "Keyboard": 28,
    "Monitor": 15
}
most_sold = max(quantity_sold,key=quantity_sold.get)
print(most_sold)
print(quantity_sold[most_sold])
```

```python
# Question 12: Find Common Products in Two Stores
store_a = {
    "Laptop": 10,
    "Mouse": 25,
    "Keyboard": 15
}

store_b = {
    "Mouse": 12,
    "Monitor": 8,
    "Keyboard": 20
}
for product in store_a:
    if product in store_b:
        print(product)
```

```python
# Question 13: Calculate Each Product's Percentage of Total Sales
sales = {
    "Laptop": 50000,
    "Mobile": 30000,
    "Tablet": 20000
}
total = sum(sales.values())
for product,amount in sales.items():
    percentage = (amount/total) * 100
    print(product,":",round(percentage,1),"%")

```

```python
# Question 14: Remove Employees from a Specific Department
employees = {
    "Amit": "IT",
    "Priya": "HR",
    "Rahul": "Finance",
    "Sneha": "HR",
    "Neha": "IT"
}
for name,department in employees.items():
    if department != 'HR':
        print(name,":",department)
```

```python
# Question 15: Find Customers Whose Spending Is Above Average
spending = {
    "C101": 5000,
    "C102": 8000,
    "C103": 3000,
    "C104": 12000,
    "C105": 7000
}
avgrage = sum(spending.values())/len(spending)
print("Avgrage:",avgrage)
for id,salary in spending.items():
    if salary > avgrage:
        print(id,":",salary )
```
