```python
# Q1. Find the third occurrence of a transaction amount
transactions = (5000, 2000, 5000, 3000, 5000, 4000, 5000)
first = transactions.index(5000)
second = transactions.index(5000, first + 1)
third = transactions.index(5000, second + 1)
print("Third occurrence index:", third)
```

```python
# Q2. Check whether all sales values are unique
sales = (1200,2300,4500,6500,8700,9000)
result = len(sales) == len(set(sales))
print(result)
```

```python
# Q3. Find the top three distinct salaries
salary = (8000,9000,2000,3000,40000,50000,70000)
result = tuple(sorted(set(salary) , reverse=True)[:3])
print(result)
```

```python
# Q4. Find the customer with the most orders
customer = (("C01",5),("C02",6),("C03",8),("C04",9),("C05",5))
result = max(customer,key=lambda item:item[1])
print(result)
```

```python
# Q5. Calculate the total of only positive adjustments
adjustments = (20,-5,15,-10,10,-3)
positive = tuple(value for value in adjustments if value > 0)
print("Total positive adjustment:", sum(positive))
```

```python
# Q6. Find the most frequent category
categories = ("Shoes","Bags","Shoes","Watches","Shoes","Bags")
most = max(categories, key=categories.count)
print("Most frequent category:", most)
print("Occurrences:", categories.count(most))
```

```python
# Q7. Group a record into separate fields
employee = (101,"Priya","Analytics",65000)
employee_id, name, * details, salary = employee
print("Employee:",name)
print("Salary:",salary)
```

```python
# Q8. Find products priced within a range
prices = (250,500,1200,2000,2500,4000)
result = tuple(price for price in prices if 500 <= price <= 2000)
print(result)
```

```python
# Q9. Compare the performance of two teams
team_a = (7000,8000,9000)
team_b = (10000,8500,8500)
avg_a = sum(team_a)/len(team_a)
avg_b = sum(team_b)/len(team_b)
better="Team A" if avg_a > avg_b else "Team B" if avg_b > avg_a else "Tie"
print("Team A average:",avg_a)
print("Team B average:",avg_b)
print("Better performance:",better)
```

```python
# Q10. Find the longest customer name
names = ("Asha","Priya Sharma","Krunali Patil","Riya")
longest=max(names,key=len)
print("Longest name:", longest)
print("Length:", len(longest))
```
