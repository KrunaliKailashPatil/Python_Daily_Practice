```python
# Question 1: Find the Employee with the Longest Name
employees = {
    "Amit": 30000,
    "Priyanka": 45000,
    "Rahul": 35000,
    "Snehal": 40000
}
largest_name = max(employees,key=len)
print(employees)
print(largest_name)
```

```python
# Question 2: Calculate Total Expenses by Category
expenses = [
    ("Food", 500),
    ("Travel", 300),
    ("Food", 200),
    ("Shopping", 800),
    ("Travel", 150)
]

total_expenses = {}
for category, amount in expenses:
    total_expenses[category] = total_expenses.get(category, 0) + amount
print(expenses)
print(total_expenses)
```

```python
# Question 3: Convert Product Names to Uppercase
products = {
    "laptop": 50000,
    "mouse": 500,
    "keyboard": 1000
}
upper_case = {}
for product,price in products.items():
    upper_case[product.upper()]=price
print(upper_case)
```

```python
# Question 4: Add a New Employee to the Dictionary
employees = {
    "Amit": 30000,
    "Priya": 45000,
    "Rahul": 35000
}
employees['Karan']=50000
employees['Tejas']=90000
print(employees)
```

```python
# Question 5: Find the Difference Between the Highest and Lowest Marks
marks = {
    "Anjali": 85,
    "Rahul": 65,
    "Meera": 95,
    "Karan": 75
}
highest = max(marks.values())
lowest = min(marks.values())
difference = highest - lowest
print("Highest marks:",highest)
print("Lowest marks:",lowest)
print("Differnece:",difference)
```

```python
# Question 6: Create a Dictionary of Even Numbers
numbers = {
    "A": 12,
    "B": 7,
    "C": 18,
    "D": 9,
    "E": 24
}
even={}
for key,number in numbers.items():
    if number % 2 == 0:
        even[key]=number
print(even)
```

```python
# Question 7: Calculate the Percentage Increase in Sales
previous_sales = {
    "Laptop": 40000,
    "Mobile": 90000
}
current_sales = {
    "Laptop": 50000,
    "Mobile": 25000
}
for products in previous_sales:
    increases = current_sales[products]-previous_sales[products]
    percentage = (increases/previous_sales[products]) * 100
    print(products,":",round(percentage,2),"%")
```

```python
# Question 8: Swap Dictionary Keys and Values
departments = {
    "Amit": "IT",
    "Priya": "HR",
    "Rahul": "Finance"
}
swap = {}
for name,department in departments.items():
    swap[department]=name
print(swap)
```

```python
# Question 9: Find Products with Zero Stock
stocks = {
    "Shirt": 20,
    "Jeans": 0,
    "Shoes": 15,
    "Cap": 0,
    "Bag": 8
}
for product,stock in stocks.items():
    if stock == 0:
        print(product,":","Is out of stock")
```

```python
# Question 10: Create a Dictionary of Word Lengths
cities = {
    "Pune": 1,
    "Mumbai": 1,
    "Nashik": 1,
    "Surat": 1
}
city_length = {}
for city in cities:
    city_length[city]=len(city)
print(city_length)
```
