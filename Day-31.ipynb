```python
# Question 1: Count the frequency of a number
numbers = [12,13,54,65,23,12,14,13,14,12]
count = 0
for num in numbers:
    if num == 12:
        count += 1
print("Frequency of 12:",count)
```

```python
# Question 2: Count the frequency of each element
numbers = [12,13,54,65,23,12,14,13,14,12]
freq = {}
for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
print(freq)
```

```python
# Question 3: Count the frequency of characters
name = "banana"
freq = {}
for char in name:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1
print(freq)
```

```python
# Question 4: Count the frequency of vowels
text = "education"
vowels = "aeiou"
count = 0
for char in text:
    if char in vowels:
        count += 1
print(count)
```

```python
# Question 5: Find the most frequent number
numbers = [1,2,3,4,5,6,1,1,1,1,1,1]
freq = {}
for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
print(freq)

most_common = max(freq,key=freq.get)
print("most frequent number:",most_common)
print("frequency:",freq[most_common])
```

```python
# Question 6: Find the least frequent number
numbers = [1,2,2,3,3,4,5,6,1,1,1,1,1,1]
freq = {}
for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
print(freq)

most_common = min(freq,key=freq.get)
print("most frequent number:",most_common)
print("frequency:",freq[most_common])
```

```python
# Question 7: Count duplicate elements
numbers = [1,2,2,3,3,4,5,6,1,1,1,1,1,1]
freq = {}
for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

duplicate = []

for num in freq:
    if freq[num] > 1:
        duplicate.append(num)
print("Duplicate element:",duplicate)
```

```python
# Question 8: Count unique elements
numbers = [1,2,2,3,3,4,5,6,1,1,1,1,1,1,7]
unique = set(numbers)
print("All numbers:",numbers)
print("Unique numbers:",len(unique))
```

```python
# Question 9: Count the frequency of words
sentence = "Python is easy and python is powerfull"
words = sentence.split()
freq = {}
for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1
print("frequency of words:",freq)
```

```python
# Question 10: Count even and odd numbers
numbers = [12,33,54,23,76,98,34,13,76,46,45]
even = 0
odd = 0
for num in numbers:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1
print("Even numbers:",even)
print("Odd numbers:",odd)
```

```python
# Question 11: Count positive and negative numbers
numbers = [12,33,54,23,76,98,34,13,-76,-46,-45,0,0,0]
positive = 0
negative = 0
zero = 0
for num in numbers:
    if num > 0:
        positive += 1
    elif num < 0:
        negative += 1
    else:
        zero += 1

print("Positive numbers:",positive)
print("Negative numbers:",negative)
print("Zero numbers:",zero)
```

```python
# Question 12: Count frequency without using a dictionary
numbers = [12,33,54,23,76,98,34,13,76,46,45]
unique = set(numbers)
for num in unique:
    print(num,"Appers",numbers.count(num),"Times")
```
