```python
# Q1. Count the frequency of numbers divisible by 5
numbers = [10,2,34,35,30,40,79,86]
count = 0
for num in numbers:
    if num % 5 == 0:
        count += 1
print("Divisible by 5:",count)
```

```python
# Q2. Count the frequency of characters at the end of a string
text = "Programming"
last_index = text[-1]
count = 0

for char in text:
    if char in last_index:
        count += 1
print(last_index)
print(count)
```

```python
# Q3. Count the frequency of numbers in a given range
numbers = [5,10,15,20,25,30,35,40]
count = 0
for num in numbers:
    if 10 <= num <= 30:
        count += 1
print(count)
```

```python
# Q4. Count the frequency of words starting with a particular letter
words = ["apple","banana","avocado","mango","apple"]
count = 0
for word in words:
    if word.startswith("a"):
        count += 1
print(count)
```

```python
# Q5. Count the frequency of repeated adjacent elements
numbers = [1,1,2,3,3,3,4,5,5]
count = 0
for i in range(len(numbers)-1):
    if numbers[i] == numbers[i+1]:
        count += 1
print("adjacent elements:",count)
```

```python
# Q6. Count the frequency of characters in reverse order
text = "apple"
freq = {}
for char in reversed(text):
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1
print(freq)
```

```python
# Q7. Count the frequency of elements in a set using a list
numbers = [1,2,3,2,4,3,2,5]
required = {2,3,4}
freq = {}

for num in required:
    freq[num] = numbers.count(num)
print(freq)
```

```python
# Q8. Find the frequency of numbers having the same last digit
numbers = [15,25,34,56,76,87,45,45]
count = 0
for num in numbers:
    if num % 10 == 5:
        count += 1

print("Frequency:",count)
```

```python
# Q9. Count the frequency of words having more than five characters
words = ["python","java","sql","excel","science"]
count = 0
for word in words:
    if len(word) > 5:
        count += 1
print("more than five character:",count)
```

```python
# Q10. Find the frequency of each element and separate them into groups
numbers = [1,2,2,3,3,3,3,4,4,5,5,5]
freq = {}

for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

once = []
twise = []
thrice = []

for num in numbers:
    if freq[num] == 1:
        once.append(num)
    elif freq[num] == 2:
        twise.append(num)
    elif freq[num] == 3:
        thrice.append(num)

print("Once:",once)
print("Twise:",twise)
print("Thrice:",thrice)
```

```python
# Q11. Count the frequency of each type of data
data = ["Banana",0.10,23,"apple",34,90.98,"pineapple"]
integer = 0
string = 0
floating = 0

for item in data:
    if isinstance(item,int):
        integer += 1
    elif isinstance(item,str):
        string += 1
    elif isinstance(item,float):
        floating += 1

print("Integer:",integer)
print("String:",string)
print("Float:",floating)
```

```python
# Q12. Find the frequency of each word and display only words occurring once
sentence = "python sql excel python powerbi pandas numpy"
words = sentence.split()
freq = {}
for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

for word in freq:
    if freq[word] == 1:
        print(word)
```
