```python
# Question 1: Display All Product Names
product = {
    "Pen": 20,
    "Pencil": 10,
    "Eraser":25,
    "Notebook":60
}
print(product)

for pro in product.keys():
    print(pro)
```

```python
# Question 2: Find the Number of Employees
Emp = {
    "E01":"Anjali",
    "E02":"Neha",
    "E03":"Nikita",
    "E04":"Krunali",
    "E05":"Manisha"
}
print(Emp)
print("Total Emp:",len(Emp))
```

```python
# Question 3: Give a Discount to One Product
product = {
    "Pen": 20,
    "Pencil": 10,
    "Eraser":25,
    "Notebook":60
}
print(product)
product["Notebook"] = 60-10
print(product)
```

```python
# Question 4: Find Customers from Mumbai
customer = {
    "Mumbai":"Anjali",
    "Pune":"Neha",
    "Delhi":"Nikita",
    "Goa":"Krunali",
    "Mumbai":"Manisha"
}

for city, cust_name in customer.items():
    if city == "Mumbai":
        print(city,":",cust_name)
```

```python
# Question 5: Create a Dictionary from Two Lists
product = ["Pen","Pencil","Notebook","Eraser"]
price = [20,10,60,25]

product_price = dict(zip(product,price))

print(product_price)
```

```python
# Question 6: Add Tax to Product Prices
prices = {
    "Book": 200,
    "Bag": 800,
    "Pen": 50
}
tax = {}
for product,price in prices.items():
    tax[product] = price * 1.05
print(tax)
```

```python
# Question 7: Find Employees Earning Below ₹30,000   
salaries = {
    "Anjali": 35000,
    "Rahul": 25000,
    "Priya": 28000,
    "Amit": 40000,
    "Sneha": 22000
}
print(salaries)

for name,salary in salaries.items():
    if salary < 30000:
        print(name,salary)
```

```python
# Question 8: Give a Default Value for a Missing Key
ratings = {
    "C101": 5,
    "C102": 4,
    "C103": 3,
    "C104": 5
}
rating = ratings.get("C106","Not Availble")
print(rating)
```

```python
# Question 9: Calculate Total Quantity in Inventory
stock = {
    "Book": 20,
    "Bag": 8,
    "Pen": 500
}

total_stock = sum(stock.values())
print("Total stock:",total_stock)
```

```python
# Question 10: Separate Pass and Fail Students
School = {
    "Anjali": 75,
    "Rahul": 32,
    "Priya": 45,
    "Amit": 28,
    "Sneha": 60
}
passed=[]
failed=[] 

for name,marks in School.items():
    if marks >= 40:
        passed.append(name)
    else:
        failed.append(name)

print("Pass Student:",passed)
print("Fail Student:",failed)
```
