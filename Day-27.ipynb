```python
# Question 1: Find Common Customers Between Two Months
jan = {"C101","C102","C103","C104","C105"}
feb = {"C103","C104","C105","C106","C107"}
common_customer = jan.intersection(feb)
print(common_customer)
```

```python
# Question 2: Find Customers Who Purchased Only in January
jan = {"C101","C102","C103","C104","C105"}
feb = {"C103","C104","C105","C106","C107"}
only_jan = jan.difference(feb)
print(only_jan)
```

```python
# Question 3: Find All Unique Products Sold by Two Stores
store_a={"Laptop","Mouse","Keyboard"}
store_b={"Mouse","Monitor","Laptop","Printer"}
unique = store_a.union(store_b)
print(unique)
```

```python
# Question 4: Find Products Available in One Store but Not Both
store_a={"Laptop","Mouse","Keyboard"}
store_b={"Mouse","Monitor","Laptop","Printer"}
unique_product=store_a.symmetric_difference(store_b)
print(unique_product)
```

```python
# Question 5: Remove Duplicate Customer IDs from a List
customer = {"C101","C102","C103","C104","C105","C102","C103","C102"}
unique_customer = set(customer)
print(unique_customer)
print("Total unique id:",len(unique_customer))
```

```python
# Question 6: Check Whether All Required Skills Are Available
required_skills={"Python","SQL","PowerBI","Excel"}
candidate_skills={"SQL","Python","PowerBI"}
if required_skills.issubset(candidate_skills):
     print("Candidate has all required skills")
else:
    print("Some skills are missing")
```

```python
# Question 7: Find Missing Required Skills
required_skills={"Python","SQL","PowerBI","Excel"}
candidate_skills={"SQL","Python","PowerBI"}
missing_skills=required_skills.difference(candidate_skills)
print(missing_skills)
```

```python
# Question 8: Check Whether Two Departments Have Any Common Employees
IT_emp={"Rahul","Ajay","Vishal","Tejas","Pravin"}
HR_emp={"Sakshi","Manisha","Vishal","Tejas","Komal"}
if IT_emp.isdisjoint(HR_emp):
    print("No common employees")
else:
    print("Common employees exist")
```

```python
# Question 9: Find Unique Website Visitors
day1={"U101","U102","U103"}
day2={"U102","U104","U105"}
day3={"U101","U105","U106"}
unique_visitors = day1.union(day2,day3)
print("Unique Visitors:",unique_visitors)
print("Total Unique Visitors:",len(unique_visitors))
```
