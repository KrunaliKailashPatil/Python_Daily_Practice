```python
# Question 1: Count Positive, Negative, and Zero Values
numbers = [10, -5, 0, 20, -3, 0, 8]
positive = 0
negative = 0
zero = 0

for num in numbers:
    if num > 0:
        positive += 1
    elif num < 0:
        negative += 1
    elif num == 0:
        zero += 1

print("Positive:",positive)
print("Negative:",negative)
print("Zero:",zero)
```

```python
# Question 2: Calculate the Running Total
num = [10,20,30,59,68,23,20,98,13]
result = []
total = 0

for i in num:
    total += i
    result.append(total)

print(result)
```

```python
# Question 3: Find the Index of Every Occurrence of a Value
number = [10,20,30,10,20,30,20]
index = []

for num in range(len(number)):
    if number[num] == 20:
        index.append(num)

print(index)
```

```python
# Question 4: Check Whether a List Is a Palindrome
number = [1,2,3,2,1]
if number == number[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
# Question 5: Remove Every Second Element
number = [1,4,3,9,1]
result = []

for num in range(0,len(number),2):
    result.append(number[num])

print(result)
```

```python
# Question 6: Find the Product of All Non-Zero Numbers
numbers = [2,0,3,4,0,5]
product = 1

for num in numbers:
    if num != 0:
        product = product * num

print(product)
```

```python
# Question 7. Replace Even Numbers with Their Squares
number = [2,3,4,5,6,7,8,9,100]
result = []

for num in number:
    if num % 2 == 0:
        result.append(num**2)
    else:
        result.append(num)

print(result)
```

```python
# Question 8. Remove All Occurrences of a Number Without Using remove()
number = [10,20,30,40,10]
target = 10
result = []

for num in number:
    if num != target:
        result.append(num)

print(result)
```

```python
# Question 10. Find the Number Closest to Zero
numbers = [-8,3,-2,7,1]
closest = numbers[0]

for num in numbers:
    if abs(num) < abs(closest):
        closest = num

print(closest)
```

```python
# Question 10. Replace Each Number with Its Sign
numbers = [10, -5, 0, 20, -8]
result = []

for num in numbers:
    if num > 0:
        result.append("*")
    elif num < 0:
        result.append("%")
    else:
        result.append("@")

print(result)
```
