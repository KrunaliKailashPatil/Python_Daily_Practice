```python
# Question 1. Remove duplicates from a list while maintaining the original order.
number = [10,20,30,40,50,60,30,50,20]
unique = []

for num in number:
    if num not in unique:
        unique.append(num)

print(unique)
```

```python
# Question 2. Find the second-largest unique number.
number = [10,20,30,40,50,60]
unique = list(set(number))
largest = max(number)
unique.remove(largest)
second_largest = max(unique)
print(second_largest)
```

```python
# Question 3. Count how many times an element appears.
number = [10,20,30,40,50,60,20,20,20,20]
count = 0

for num in number:
    if num == 20:
        count += 1
print(count)
```

```python
# Question 4: Combine two lists and remove duplicate values while maintaining their order.
list1=[10,20,30,30,40,40,50]
list2=[50,50,60,70,80,90,100]

list1.extend(list2)
result = []

for num in list1:
    if num not in result:
        result.append(num)

print(result)
```

```python
# Question 5: Print only the numbers that appear more than once, without printing duplicates in the result.
number = [10,20,30,40,50,60,20,20,20,20]
repeat = []
for num in number:
    if number.count(num) > 1 and num not in repeat:
        repeat.append(num)

print(repeat)
```

```python
# Question 6: Create a copy, add 60 to the copy, and print both lists.
number = [10,20,30,40]

new = number.copy()
new.append(50)

print(number)
print(new)
```

```python
# Question 7: Remove all elements from the list and check whether it is empty.
number = [10,20,30,40,50,60,20,20,20,20]
number.clear()

if len(number) == 0:
    print("List is empty")
else:
    print("List is not empty")

print(number)
```

```python
# Question 8: Find the largest and smallest numbers without using max() or min().
number = [10,20,30,40,50,60,20,20,20,20]

big = small = number[0]

for num in number:
    if num > big:
        big = num
    if num < small:
        small = num

print("Largest:",big)
print("Smallest:",small)
```

```python
# Question 9: Move negative numbers to the beginning.
number = [-11,10,340,56,-4,-6,-3,40,-9,-23,78,98,56]
negative = []
postive = []

for num in number:
    if num < 0:
        negative.append(num)
    else:
        postive.append(num)

negative.extend(postive)
print(negative)
```

```python
# Question 10: Find the intersection of multiple lists
list1 = [10, 20, 30, 40, 50]
list2 = [20, 30, 50, 60]
list3 = [10, 20, 30, 50]

result = []

for num in list1:
    if num in list2 and num in list3:
        if num not in result:
            result.append(num)

print(result)
```
