```python
# Question 1: Find Unique Employee Departments
department = {'HR','IT','Sales','Finance','IT','Sales'}
unique_dept = set(department)
print(unique_dept)
```

```python
# Question 2: Remove a Specific Product
products = {"laptop","Mosue","Keyboard","Monitor"}
products.remove("laptop")
print(products)
```

```python
# Question 3: Check Whether a Customer Exists
customers={"C101","C102","C103","C104","C105"}
if "C104" in customers:
    print("Customer exists")
else:
    print("Customer not found")
```

```python
# Question 4: Find Unique Numbers Greater Than 50
numbers = [20,34,56,23,12,87,90,12,89,56]
result = {num for num in numbers if num > 50}
print(result)
```

```python
# Question 5: Find Common Skills Between Two Employees
emp_a = {"Python","SQL","Kotlin","Java","Excel"}
emp_b = {"SQL","Python","Power BI","Java","C++"}
common = emp_a&emp_b
print(common)
```

```python
# Question 6: Find Products Not Available in the Warehouse
required_products = {"Laptop","Mouse","Keyboard","Printer","Monitor"}
available = {"Laptop","Mouse"}
missing_product = required_products-available
print(missing_product)
```

```python
# Question 7: Find the Highest Unique Score
scores = {78,98,34,24,90,24,56,78}
highest_score = max(scores)
print(highest_score)
```

```python
# Question 8: Find the Lowest Unique Sales Value
scores = {78,98,34,24,90,24,56,78}
highest_score = min(scores)
print(highest_score)
```

```python
# Question 9: Remove All Records from a Set
scores = {78,98,34,24,90,24,56,78}
print(scores)
scores.clear()
print(scores)
```

```python
# Question 10: Count Unique Customer Names Starting with A
customers = ["Amit","Priya","Rahul","Vishal","Krunali","Manisha","Dipali"]
unique_custo = set(customers)
result = {name for name in unique_custo if name.startswith("A")}
print(result)
print(len(result))
```
