```python
# Q2. Add a New Product
products = {
    "Mouse": 500,
    "Monitor": 8000,
    "Laptop": 45000
}
products["Keyboard"]=1200
print(products)
```

```python
# Q3. Update Employee Salary
employee_salary = {
    "Rahul": 25000,
    "Priya": 35000,
    "Amit": 28000
}
employee_salary["Rahul"]=30000
print(employee_salary)
```

```python
# Q4. Find Total Sales
sales = {
    "Shoes": 5000,
    "Bags": 3000,
    "Watches": 2500
}
total_sales=sum(sales.values())
print("Total Sales:",total_sales)
```

```python
# Q5. Find the Highest Scorer
student_marks = {
    "Anjali": 85,
    "Rahul": 92,
    "Priya": 78,
    "Amit": 88,
    "Sneha": 95
}
top_student=max(student_marks, key=student_marks.get)
print("Top Student:",top_student)
print("Highest Marks:",student_marks[top_student])
```

```python
# Q6. Count Product Categories
categories = [
    "Clothing",
    "Shoes",
    "Clothing",
    "Electronics",
    "Shoes",
    "Clothing"
]
category_count = {}
for category in categories:
    category_count[category]=category_count.get(category, 0) + 1
print(category_count)
```

```python
# Q7. Find Customers Who Have Not Paid
payments = {
    "Amit":"Paid",
    "Priya":"Pending",
    "Rahul":"Pending",
    "Sneha":"Paid",
    "Karan":"Pending"
}
for customer, status in payments.items():
    if status == "Pending":
        print(customer)
```

```python
# Q8. Calculate Average Marks
student_marks = {
    "Anjali": 80,
    "Rahul": 90,
    "Priya": 70,
    "Amit": 100
}
average_marks = sum(student_marks.values()) / len(student_marks)
print("Average Marks:", average_marks)
```

```python
# Q9. Filter High-Value Customers
customer_spending = {
    "Amit": 2500,
    "Priya": 7000,
    "Rahul": 5500,
    "Sneha": 3200,
    "Karan": 9000
}
for customer, amount in customer_spending.items():
    if amount >= 5000:
        print(customer, amount)
```

```python
# Q10. Create a Discounted Price Dictionary
product_prices = {
    "Shoes": 2000,
    "Bag": 1500,
    "Watch": 3000
}
discounted_prices = {}
for product, price in product_prices.items():
    discounted_prices[product] = price * 0.90
print(discounted_prices)
```
