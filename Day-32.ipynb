```python
# Q1. Count how many times a specific character appears at an even index
text = "bananaapple"
count = 0
for i in range(0, len(text)):
    if text[i] == 'a':
        count += 1

print("Frequency:", count)
```

```python
# Q2. Find numbers that appear exactly three times
numbers = [10,20,10,20,40,30,10,20,40,30,20,70]
freq = {}
for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

result = []

for num in freq:
    if freq[num] == 3:
        result.append(num)

print(result)
```

```python
# Q3. Count uppercase and lowercase character frequency
text = "PythonProgramming"
upper = 0
lower = 0

for char in text:
    if char.isupper():
        upper += 1
    elif char.islower():
        lower += 1

print("Upper:",upper)
print("Lower:",lower)
```

```python
# Q4. Count frequency of numbers greater than 10
numbers = [8,9,45,38,98,39,29,9,8,5,3,2,0]
count = 0

for num in numbers:
    if num > 10:
        count += 1

print("Greater then 10 number:",count)
```

```python
# Q5. Find the frequency of each digit in a number
numbers = 123456654321
text = str(numbers)
freq = {}

for digit in text:
    if text in freq:
        freq[digit] += 1
    else:
        freq[digit] = 1
print(freq)
```

```python
# Q7. Find the frequency of each element in a tuple
fruits = ("Apple","Banana","Pineapple","Mango","Apple")
freq = {}

for fruit in fruits:
    if fruit in freq:
        freq[fruit] += 1
    else:
        freq[fruit] = 1
print(freq)
```

```python
# Q8. Find the frequency of elements at odd positions
numbers = [10,20,30,40,50,60,70,80,80,90]
freq = {}

for i in range(1,len(numbers),2):
    num = numbers[i]

    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
print(freq)
```

```python
# Q9. Find the frequency of elements in a nested list
numbers = [[1,2,3],[4,2,6],[2,3,4]]
freq = {}

for sublist in numbers:
    for num in sublist:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1
print(freq)
```

```python
# Q10. Find the frequency of each character without counting spaces
text = "Data Science"
freq = {}
for char in text:
    if char != " ":
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1
print(freq)
        
```

```python
# Q11. Find the frequency of each number and calculate its percentage
numbers = [10,20,30,40,50,60,10,20,30,40,40,40,40]
freq = {}

for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

total = len(numbers)

for num in freq:
    percentage = (freq[num] / total) * 100
    print(f"{num}: {round(percentage, 1)}%")
```

```python
# Q12. Find the non-repeating character
text = "aabbccde"
freq = {}

for char in text:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1

for char in text:
    if freq[char] == 1:
        print("Non-repeating charater:",char)
```
