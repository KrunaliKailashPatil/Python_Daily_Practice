```python
# Question 1: Find the Product with the Lowest Price
prices = {
    "Laptop": 50000,
    "Mouse": 500,
    "Keyboard": 1000,
    "Monitor": 8000
}
cheapest_price = min(prices,key=prices.get)

print(prices)
print(prices[cheapest_price])
```

```python
# Question 2: Count Orders by Status
orders = {
    "O101": "Delivered",
    "O102": "Pending",
    "O103": "Delivered",
    "O104": "Cancelled",
    "O105": "Pending",
    "O106": "Delivered"
}
status_count = {}

for status in orders.values():
    status_count[status] = status_count.get(status,0) + 1
print(status_count)
```

```python
# Question 3: Find the Highest-Paid Employee
salaries = {
    "Amit": 25000,
    "Priya": 42000,
    "Rahul": 35000,
    "Sneha": 50000
}
highest_payment = max(salaries,key=salaries.get)
print(salaries)
print(highest_payment)
```

```python
# Question 4: Calculate Total Sales 
sales = {
    "Shoes": 12000,
    "Bags": 8000,
    "Watches": 15000,
    "Belts": 3000
}
total_sales = sum(sales.values())
print(sales)
print(total_sales)
```

```python
# Question 5: Find Products Below ₹1,000
products = {
    "Mouse": 500,
    "Keyboard": 1200,
    "USB Cable": 250,
    "Headphones": 1500,
    "Pen Drive": 800
}
for product,price in products.items():
    if price < 1000:
        print(product,":",price)
```

```python
# Question 6: Update Stock After a Sale
inventory = {
    "Notebook": 50,
    "Pen": 100,
    "Pencil": 80
}
inventory["Notebook"] = 50 + 12
print(inventory)
```

```python
# Question 7: Find Students Who Scored Above 75
marks = {
    "Anjali": 82,
    "Rahul": 67,
    "Meera": 91,
    "Karan": 39,
    "Pooja": 88
}
for name,mark in marks.items():
    if mark >= 40:
        print("Passed-",name,":",mark)
    else:
        print("Failed-",name,":",mark)
```

```python
# Question 8: Calculate Revenue for Each Product
quantity = {
    "Laptop": 2,
    "Mouse": 10,
    "Keyboard": 5
}

price = {
    "Laptop": 50000,
    "Mouse": 500,
    "Keyboard": 1000
}
revenue = {}
for product in quantity:
    revenue[product] = quantity[product] * price[product]

print(revenue)
```

```python
# Question 9: Find Employees Missing Their Email Address
employees = {
    "Amit": "amit@gmail.com",
    "Priya": None,
    "Rahul": "rahul@gmail.com",
    "Sneha": None
}
for name, email in employees.items():
    if email is None:
        print(name)
```

```python
# Question 10: Give a Bonus to Employees
salaries = {
    "Amit": 25000,
    "Priya": 40000,
    "Rahul": 28000,
    "Sneha": 35000
}
updated_salaries = {}
for name, salary in salaries.items():
    if salary < 30000:
        updated_salaries[name] = salary + 2000
    else:
        updated_salaries[name] = salary
print(updated_salaries)
```
