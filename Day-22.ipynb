```python
# Question 1: Find the Second-Highest Sales Amount
sales = {
    "Anjali": 65000,
    "Rahul": 85000,
    "Priya": 92000,
    "Amit": 75000,
    "Sneha": 78000
}
sorted_sales = sorted(sales.items(),key=lambda item:item[1],reverse=True)
print(sales)
print(sorted_sales)
print(sorted_sales[1][1])
```

```python
# Question 2: Calculate the Percentage of Each Expense
expenses = {
    "Rent": 10000,
    "Food": 5000,
    "Travel": 3000,
    "Shopping": 2000
}
total = sum(expenses.values())

for category,amount in expenses.items():
    percentage = (amount/total) * 100
    print(f"{category}:{percentage:.1f}%")
```

```python
# Question 5: Replace Missing Ratings with a Default Value
ratings = {
    "C101": 5,
    "C102": None,
    "C103": 4,
    "C104": None,
    "C105": 3
}
for id, rating in ratings.items():
    if rating is None:
        ratings[id] = 0

print(ratings)
```

```python
# Question 6: Group Employees by Department
employees = {
    "Anjali": "IT",
    "Rahul": "IT",
    "Priya": "HR",
    "Amit": "Finance",
    "Sneha": "HR"
}
department_groups = {}

for employee, department in employees.items():
    if department not in department_groups:
        department_groups[department] = []

    department_groups[department].append(employee)
print(department_groups)
```

```python
# Question 7: Find the Difference Between Two Months' Sales
january = {
    "Laptop": 50000,
    "Mobile": 40000,
    "Tablet": 20000
}
february = {
    "Laptop": 60000,
    "Mobile": 35000,
    "Tablet": 25000
}
for product in january:
    difference = february[product] - january[product]
    print(f"{product}:{difference}")
```

```python
# Question 8: Apply a Bonus to Product Prices
products = {
    "Pen": 50,
    "Book": 200,
    "Bag": 600
}
for product in products:
    products[product] = products[product] + 100
print(products)
```

```python
# Question 9: Calculate Attendance Percentage
attendance = {
    "Anjali": 18,
    "Rahul": 15,
    "Priya": 19,
    "Amit": 12
}
total_days = 20

for student,attended_days in attendance.items():
    percentage = (attended_days/total_days) * 100
    print(f"{student}:{percentage:.1f}%")
```
