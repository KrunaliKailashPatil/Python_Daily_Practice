```python
# Q1. Find the longest word in a sentence.
sentence = "Python programming is very intresting"
word = sentence.split()
longest = ""
for word in word:
    if len(word)>len(longest):
        longest=word
print("Longest word:",longest)
print("Length:",len(longest))

```

```python
# Q2. Remove all special characters from a string.
text = "Hello@Python#123! Welcome$"
result =""
for char in text:
    if char.isalnum() or char.isspace():
        result += char
print(result)
```

```python
# Q3. Find the number of vowels and consonants in a string.
text = "Python developer 2026"
vowels = "aeiou"
vowel_count = 0
consonant_count = 0

for char in text.lower():
    if char.isalpha():
        if char in vowels:
            vowel_count += 1
        else:
            consonant_count += 1
            
print("Vowels:",vowel_count)
print("Consonant:",consonant_count)
```

```python
# Q4. Rotate a list to the right by two positions.
numbers = [10,20,30,40,50,60,70,80]
rotated = numbers[-2:] + numbers[:-2]
print(rotated)
```

```python
# Q5. Find the pairs whose sum equals a target value.
numbers = [2,3,5,6,7,9,8,1,2,3]
target = 9
pairs = []
for i in range(len(numbers)):
    for j in range(i+1,len(numbers)):
        if numbers[i] + numbers[j] == target:
            pairs.append((numbers[i],numbers[j]))
print(pairs)
```

```python
# Q6. Move all zeros to the end of a list.
numbers = [0,5,9,3,0,4,7,2,0,0]
non_zero = []
zeros = []
for num in numbers:
    if num == 0:
        zeros.append(num)
    else:
        non_zero.append(num)
result = non_zero + zeros
print("Non zeros number:",non_zero)
print("Zeros number:",zeros)
print(result)
```

```python
# Q7. Sort a tuple of records according to marks.
student = (("Neha",80),("Manisha",89),("Dipali",78),("Amit",88))
result = tuple(sorted(student,key=lambda x:x[1],reverse=True))
print(result)
```

```python
# Q8. Find the common subjects studied by all students.
stud1=("Python","SQL","ML")
stud2=("PowerBI","SQL","Python")
stud3=("Excel","SQL","PowerBI","Python")
common = set(stud1) & set(stud2) & set(stud3)
result=tuple(sorted(common))
print(result)
```

```python
# Q9. Unpack a tuple and calculate the average marks.
student = ("Neha",89,87,80)
name,m1,m2,m3 = student
total = m1+m2+m3
average = total / 3
print("student name:",name)
print("Total marks:",total)
print("average:",int(average))
```

```python
# Q10. Merge two dictionaries and add values for common keys.
sales1 = {"Laptop":7,"Mobile":4,"Tablet":5}
sales2 = {"Laptop":15,"Mobile":10,"Tablet":8}
result=sales1.copy()
for product,sales in sales2.items():
    if product in result:
        result[product] += sales
    else:
        result[product] = sales
print(result)
```
