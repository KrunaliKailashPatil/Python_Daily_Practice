```python
# Q1. Find the frequency of each character in a string.
name = "banana"
char = {}
for each in name:
    if each in char:
        char[each]+=1
    else:
        char[each]=1
print(char) 
```

```python
# Q2. Find the first non-repeating character.
text = "swiss"
frequency = {}
for char in text:
    frequency[char]=frequency.get(char,0)+1
result = None
for char in text:
    if frequency[char]==1:
        result=char
        break
if result:
    print("First non-repeting character:",result)
else:
    print("No unique character")
```

```python
# Q3. Check whether two strings are anagrams.
str1 = "listen"
str2 = "silent"
str1 = str1.replace(" ","").lower()
str2 = str2.replace(" ","").lower()
if sorted(str1)==sorted(str2):
    print("The string is anagrams")
else:
    print("The string is not anagrams")
```

```python
# Q4. Find the second-largest unique number in a list.
numbers = [10,20,30,40,50]
unique = list(set(numbers))
unique.sort(reverse=True)
if len(unique) >= 2:
    print("Second largest numbers:",unique[1])
else:
    print("Second largest number i not available")
```

```python
# Q5. Separate even and odd numbers into two lists.
even_odd = [10,15,20,25,30,35,40,45,50,55,60,65,70]
even = []
odd = []
for num in even_odd:
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)
print("Even list:",even)
print("Odd list:",odd)
```

```python
# Q6. Find common elements between two lists without duplicates.
list1=[10,20,30,40,50]
list2=[40,50,60,70,80]
common=[]
for num in list1:
    if num in list2 and num not in common:
        common.append(num)
print(f"Common list:{common}")
```

```python
# Q7. Find the student with the highest marks using a tuple.
student = (("Neha",98),("Krish",78),("Rahul",89),("Tejas",70))
topper = ""
highest = 0
for name,marks in student:
    if marks > highest:
        highest = marks
        topper = name
print("Topeer:",topper)
print("Highest Marks:",highest)
```

```python
# Q8. Convert a tuple into a dictionary.
student = (("Neha",98),("Krish",78),("Rahul",89),("Tejas",70))
result = dict(student)
print(result)
```

```python
# Q9. Find the student who scored more than 80 marks.
student = (("Neha",98),("Krish",78),("Rahul",89),("Tejas",70))
result = {}
for name,marks in student:
    if marks > 80:
        result[name]=marks
print(result)
```

```python
# Q10. Find the product with the highest sales.
sales = {"Mobile":30000,"Laptop":40000,"Tablet":25000,"Keyboard":15000}
highest_sales = 0
product_name = ""
for product,price in sales.items():
    if price > highest_sales:
        highest_sales = price
        product_name = product
print("Highest selling products:",product_name)
print("Sales:",highest_sales)
```

```python
# Q11. Find the union, intersection and difference of two sets.
set1={10,20,30,40}
set2={30,40,50,60}
print("Union:",set1.union(set2))
print("Intersection:",set1.intersection(set2))
print("Difference:",set1.difference(set2))
```

```python
# Q13: Find Duplicate Transaction IDs
transactions = [101,102,103,104,105,106,107,108,102,103,104]
unique_id = set()
duplicates_id = set()
for transcation in transactions:
    if transcation in unique_id:
        duplicates_id.add(transcation)
    else:
        unique_id.add(transcation)
print(duplicates_id)
```
