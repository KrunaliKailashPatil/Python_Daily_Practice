```python
# Q1. Find the position of a value
emp_id = (101,102,103,104,105,106,107,108,109,110)
position = emp_id.index(106)
print(position)
```

```python
# Q2. Count repeated values
prices = (101,100,200,400,300,200,350,233,560)
repeat = prices.count(200)
print(repeat)
```

```python
# Q3. Find the second-highest unique value
prices = (101,100,200,400,300,200,350,233,560)
unique = sorted(set(prices))
second_high = unique[-2]
print(second_high)
```

```python
# Q4. Reverse a tuple without a loop
prices = (101,100,200,400,300,200,350,233,560)
rev = prices[::-1]
print(rev)
```

```python
# Q5. Extract values from a tuple
number = (10,20,30,40,50,60,40,20,30,10)
middle = number[3:6]
print(middle)
```

```python
# Q6. Find all positions of a repeated value
data = (12,23,12,23,12,44,56,76,12)
position = []

for i,value in enumerate(data):
    if value == 12:
        position.append(i)
print(position)
```

```python
# Q7. Separate even and odd numbers
data = (10,11,121,31,4,54,65,78,34,5,42,36,78)
even = tuple(n for n in data if n%2==0)
odd = tuple(n for n in data if n%2!=0)
print("Even:",even)
print("Odd:",odd)
```

```python
# Q8. Find the best-performing sales region
sales = (
    ("Pune", 85000),
    ("Mumbai", 120000),
    ("Nashik", 65000),
    ("Nagpur", 95000)
)
best = max(sales)
lowest = min(sales)
difference = best[1] - lowest[1]

print("Best region:", best)
print("Highest sales:", best)
print("Lowest sales:", lowest)
print("Sales difference:", difference)
```

```python
# Q9. Calculate product-wise revenue
products = (
    ("Shoes", 10, 2000),
    ("Bags", 15, 1200),
    ("Watches", 8, 3000),
    ("Shirts", 20, 800)
)

for product, quantity, price in products:
    revenue = quantity * price
    print(f"{product}: ₹{revenue}")
```

```python
# Q10. Compare monthly sales against a target
monthly_sales = (
    ("January", 85000),
    ("February", 110000),
    ("March", 100000),
    ("April", 75000),
    ("May", 125000)
)

target = 100000
for month, actual_sales in monthly_sales:
    percentage = (actual_sales / target) * 100
    if actual_sales >= target:
        status = "Target Achieved"
    else:
        status = "Target Not Achieved"
    print(f"{month}: {percentage:.1f}% - {status}")
```
